```python
students=[]
students.append("John") #append stores the value given in list at the end of the list 
students.append("Alice") #insert stores the value given in list at the index specified in the list
students.insert(1,"Bob") #extend stores the value given in list at the end of the list but it can take multiple values in a list
students.append(["Charlie","David"])
students.extend(["Charlie","David"])
print(students)
s=students.pop()
print(s)
students.pop(1)
print(students)
del students[1]

print(students)
s1=students.remove("John")
print(s1)
print(students)
students.clear()
print(students)
```
