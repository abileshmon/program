```python
#online shopping bill
prod=int(input("enter no. of prod:"))
i=0
product_list=[]
high_price=0
low_price=None
high_prod=None
low_prod=None
total_prod=0
while(prod>0):
    print("enter the details of product",(i+1))
    i+=1
    product_id=int(input("enter product id"))
    product_name=input("enter product name")
    product_price=float(input("enter product price"))
    if high_price<product_price:
        high_price=product_price
        high_prod=product_name
    if low_price is None  or low_price > product_price:
        low_price=product_price
        low_prod=product_name    
    product_quantity=int(input("enter product quantity"))
    total_prod+=product_quantity
    total=product_price*product_quantity
    product_list.append((product_id,product_name,product_price,product_quantity,total))
    prod-=1
print("----------------------------------------------------------------------")
print("{:<6} {:<15} {:<10} {:<5} {:<7}".format("ID", "name", "price", "qty", "total"))
print("----------------------------------------------------------------------")
for product in product_list:
    print("{:<6} {:<15} {:<10} {:<5} {:<7}".format(product[0],product[1],product[2],product[3],product[4]))
grand_total=0
for product in product_list:
    grand_total=grand_total+product[4]
print("grand total:",grand_total)
print("expensive product:",high_prod)
print("cheap product:",low_prod)
if grand_total> 10000.0:
    discount=(10/100)*grand_total
    grand_total=grand_total-discount
    print("total after discount:",grand_total)
gst=(5/100)*grand_total
gst_total=grand_total+gst    
print("total after gst:",gst_total)
id=int(input("enter prod id"))
for product in product_list:
    if id == product[0]:
        print("product of",id,"number is:",product[1])
        break
    else:
        print("product not found")

print("total number of product purchased:",total_prod)  
            
    
```
