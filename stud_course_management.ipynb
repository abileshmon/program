```python
students = [
[101, "Alice", "Python", 15000, "Active"],
[102, "Bob", "Java", 12000, "Active"],
[103, "Charlie", "Full Stack", 20000, "Inactive"],
[104, "David", "Python", 15000, "Active"]
]
while(True):
    print("=========STUDENT COURSE MANAGEMENT=========")
    print("1. Add New Student")
    print("2. Insert Student at Specific Position")
    print("3. Import Students from Another Branch")
    print("4. View All Students")
    print("5. Search Student")
    print("6. Update Student Course")
    print("7. Remove Student by Name")
    print("8. Remove Last enrolled Student")
    print("9. Backup Student records")
    print("10. Count python students")
    print("11. sort students by name")
    print("12 reverse students list")
    print("13. find student position")
    print("14. display first five students")
    print("15. display last three students")
    print("16. delete student using index")
    print("17. check student availability")
    print("18. clear all student records")
    print("19. display total number of students")
    print("20. Exit")
    choice = input("Enter your choice: ")
    if choice == "1":
      id=int(input("enter student id"))
      name=input("enter student name")
      course=input("enter course name")
      course_fee=int(input("enter course fees"))
      status=input("enter course status")
      new_stud=[id,name,course,course_fee,status]
      students.append(new_stud)
      print("Student Added Successfully")
    elif choice=="2":
      pos=int(input("enter position to store"))
      id=int(input("enter student id"))
      name=input("enter student name")
      course=input("enter course name")   
      course_fee=int(input("enter course fees"))
      status=input("enter course status")
      new_stud=[id,name,course,course_fee,status]
      students.insert(pos,new_stud)
      print("Student Inserted Successfully")
    elif choice =="3":
      branch_students = [
[105, "Eva", "Python", 15000, "Active"],
[106, "Frank", "Java", 12000, "Active"]
]
      students.extend(branch_students)
      print("students imported successfully")
    elif choice=="4":
      print(" ID   Name   course    fee    status")
      for i in students:
        print(f"{i[0]}   {i[1]}   {i[2]}   {i[3]}    {i[4]}")
    elif choice=="5":
      c=0
      name=input("enter a name")
      for i in students:
        if i[1].lower() == name:
          print("student id:",i[0])
          print("name:",i[1])
          print("course:",i[2])
          print("fees:",i[3])
          print("status:",i[4])
          c+=1
      if c==0:
        print("student not found")
    elif choice=="6":
      name=input("enter the name")
      new_course=input("enter new course")
      c=0
      for i in students:
        if i[1].lower() == name:
          i[2]=new_course
          print(students)
          c+=1
      if c==0:
        print("studennt not found")    
    elif choice=="7":
      c=0
      name=input("enter a name")
      for i in students:
        if i[1].lower()==name:
          students.remove(i)
          print(students)
          c+=1
      if c==0:
        print("student not found")    
    elif choice=="8":
      s=students.pop()
      print("\tREMOVED STUDENT\t")
      print("student id:",s[0])
      print("name:",s[1])
      print("course:",s[2])
      print("course fee:",s[3])
      print("status:",s[4])
    elif choice=="9":
      students_backup=[]
      students_backup=students.copy()
      print("backup created successfully.")
    elif choice=="10":
      c=0
      for i in students:
        if i[2]=="Python":
          c+=1
      print(f"total students studying python:{c}")          
    elif choice=="11":
      print(students)                      #not completed
    elif choice=="12":
      students.reverse()
      print(students) 
    elif choice=="13":
      name=input("enter a name")
      c=0
      count=0
      for i in students:
        c+=1
        if i[1].lower()==name:
          count+=1
          print("position is:",c)  
      if count==0:
        print("student not found")
    elif choice=="14":
      print(students[0:5])
    elif choice=="15":
      print(students[3:]) 
    elif choice=="16":
      index=int(input("enter index:"))
      if index<len(students):
        del students[index]
      else:
        print("index invalid")
    elif choice =="17":
      name=input("enter a name:")
      c=0
      for i in students:
        if name.lower() == i[1].lower():
          print("student available")
          c+=1  
      if c==0:
        print("student not available")
    elif choice=="18":
      students.clear()
      print("All students record deleted successfully")
    elif choice=="19":
      length=len(students)
      print("total students:",length)
    elif choice=="20":
      print("thankyou")
      print("student course management closed successfully")
      break
    else:
      print("invalid choice try  again")
```
